```python
print("-----task1----")

import re

# Six sample inputs
samples = [
    "AI, search!",
    "I can't log in.",
    "help@example.com",
    "CS4944 version 2.1",
    "",
    "یہ ایک مثال ہے"
]

for text in samples:

    # Method 1: Whitespace splitting
    whitespace_tokens = text.split()

    # Method 2: Unicode word-like regex
    regex_tokens = re.findall(r"\b\w+\b", text, flags=re.UNICODE)

    print("\nOriginal Text:", repr(text))

    print("Whitespace Tokens:", whitespace_tokens)
    print("Whitespace Count:", len(whitespace_tokens))

    print("Regex Tokens:", regex_tokens)
    print("Regex Count:", len(regex_tokens))


# Optional: Model tokenizer for two English examples
print("\n--- Model Tokenizer ---")

try:
    from sentence_transformers import SentenceTransformer

    model = SentenceTransformer(
        "sentence-transformers/all-MiniLM-L6-v2",
        device="cpu"
    )

    for text in samples[:2]:
        tokens = model.tokenizer.tokenize(text)
        print("Text:", text)
        print("Model Tokens:", tokens)
        print("Model Token Count:", len(tokens))

except Exception as e:
    print("Model tokenizer not available:", e)
```
